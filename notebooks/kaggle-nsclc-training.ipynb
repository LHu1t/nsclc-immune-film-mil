```python
# Checkpoint utilities
import json
import numpy as np

CHECKPOINT_FILE = "/kaggle/working/checkpoint.json"

def _json_default(obj):
    """Convert numpy scalar/array types to native Python types for json.dump."""
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    raise TypeError(f"Object of type {type(obj).__name__} is not JSON serializable")

def save_checkpoint(fold_results: list, completed_fold: int):
    """Save progress after each fold completes."""
    checkpoint = {
        "completed_fold": completed_fold,
        "fold_results":   fold_results,
    }
    with open(CHECKPOINT_FILE, "w") as f:
        json.dump(checkpoint, f, indent=2, default=_json_default)
    print(f"Checkpoint saved after fold {completed_fold}")

def load_checkpoint():
    """Load checkpoint if it exists, returns (fold_results, start_fold)."""
    if Path(CHECKPOINT_FILE).exists():
        with open(CHECKPOINT_FILE) as f:
            cp = json.load(f)
        start_fold = cp["completed_fold"] + 1
        print(f"Resuming from fold {start_fold} "
              f"({cp['completed_fold']} folds already completed)")
        return cp["fold_results"], start_fold
    return [], 0

print("Checkpoint utilities ready.")
```

## Main Training

```python
# Import training components
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from sklearn.model_selection import KFold
from pathlib import Path

from train_film_mil import (
    load_metadata,
    FiLMDataset,
    FiLMMILModel,
    CompositeLoss,
    run_epoch,
    compute_gene_pccs,
    compute_panel_pcc,
    compute_auc,
    compute_panel_head_pcc,
    APM_GENES,
    TIS_GENES,
    AGGREGATOR_CHOICES,
    preload_paths,
    configure_feature_cache,
)

print("All imports successful.")
```

```python
# Load and preprocess metadata
df, gene_cols, clinical_cols, y_means, y_stds, panel_idx = load_metadata(METADATA_CSV)

gene_symbol_to_idx = {}
for i, g in enumerate(gene_cols):
    symbol = g.replace("_fpkm_uq", "")
    gene_symbol_to_idx[symbol] = i

# Build gene symbol > index map
gene_symbol_to_idx = {}
for i, g in enumerate(gene_cols):
    symbol = g.replace("_fpkm_uq", "")
    gene_symbol_to_idx[symbol] = i

panel_gene_fallback_idx = {
    "APM": [
        gene_symbol_to_idx[g]
        for g in APM_GENES
        if g in gene_symbol_to_idx
    ],
    "TIS": [
        gene_symbol_to_idx[g]
        for g in TIS_GENES
        if g in gene_symbol_to_idx
    ],
}

feature_dirs = {"LUAD": LUAD_FEATURES, "LUSC": LUSC_FEATURES}

# Fixed 80/20 train-dev / test split
rng       = np.random.default_rng(94)
all_sids  = df["submitter_id"].unique()
test_sids = set(rng.choice(all_sids, size=int(0.2 * len(all_sids)), replace=False))
dev_sids  = [s for s in all_sids if s not in test_sids]

df_test = df[df["submitter_id"].isin(test_sids)].reset_index(drop=True)
df_dev  = df[df["submitter_id"].isin(dev_sids)].reset_index(drop=True)

print(f"Total samples : {len(df)}")
print(f"Dev set       : {len(df_dev)}")
print(f"Test set      : {len(df_test)}")
print(f"Gene targets  : {len(gene_cols)}")
#print(f"Subtypes      : LUAD={( df['cancer_type']=='LUAD').sum()}, LUSC={(df['cancer_type']=='LUSC').sum()}")
```

```python
# Main training loop with per-fold checkpointing
import os

# Choose the aggregator for this run. See AGGREGATOR_CHOICES for the full
# list ("abmil", "clam", "transmil", "graph"). Weights/results/checkpoint
# below are all kept in an aggregator-specific location, so you can rerun
# this cell with a different AGGREGATOR without overwriting a previous
# run's saved model -- handy for picking which architecture's weights to
# later load for inference on an external dataset.
AGGREGATOR = "abmil"   # one of AGGREGATOR_CHOICES
# Aggregator hyperparameters (only the ones relevant to AGGREGATOR matter;
# defaults match train_film_mil.py's FiLMMILModel defaults).
AGG_HIDDEN_DIM = 256
GRAPH_K = 8
GRAPH_LAYERS = 2
TRANSMIL_HEADS = 8
TRANSMIL_LAYERS = 2
TRANSMIL_DIM_FEEDFORWARD = 512

print(f"Folds:{N_FOLDS}, Patience: {PATIENCE}, Aggregator: {AGGREGATOR}")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Aggregator-specific output dir and checkpoint file, so switching
# AGGREGATOR and rerunning this cell never overwrites a previous run's
# saved weights or resumes from the wrong aggregator's checkpoint.
AGG_OUTPUT_DIR = f"{OUTPUT_DIR}/{AGGREGATOR}"
os.makedirs(AGG_OUTPUT_DIR, exist_ok=True)
CHECKPOINT_FILE = f"/kaggle/working/checkpoint_{AGGREGATOR}.json"

# Configure + preload the feature cache -- ONLY for the fixed test set.
#
# Train/val deliberately do NOT use the cache: DataLoader spawns
# num_workers separate PROCESSES, and any caching a worker does after
# being forked lives in that worker's own private copy-on-write memory,
# not shared with the other workers. Caching the large, every-epoch
# dev set that way can multiply effective RAM usage by num_workers and
# trigger OS-level swapping -- far slower than the plain uncached reads
# it was meant to replace. The test set is small and evaluated only
# once per fold, and is warmed here in the MAIN process before any
# workers are forked, so it's shared safely via copy-on-write.
FEATURE_CACHE_MAX_GB = 18
FEATURE_CACHE_DTYPE = "float16"
configure_feature_cache(max_gb=FEATURE_CACHE_MAX_GB, dtype=FEATURE_CACHE_DTYPE)
_test_paths = [r["h5_path"] for r in
               FiLMDataset(df_test, feature_dirs, gene_cols, clinical_cols).records]
preload_paths(_test_paths)

kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=90)

# Resume from checkpoint if available
fold_results, start_fold = load_checkpoint()

for fold, (train_idx, val_idx) in enumerate(kf.split(df_dev)):

    # Skip already-completed folds
    if fold < start_fold:
        print(f"Skipping fold {fold} (already completed)")
        continue

    print(f"\n{'='*60}")
    print(f"FOLD {fold} / {N_FOLDS - 1}")
    print(f"{'='*60}")

    df_train = df_dev.iloc[train_idx].reset_index(drop=True)
    df_val   = df_dev.iloc[val_idx].reset_index(drop=True)

    # Datasets. use_feature_cache=False for train/val (bypass the cache --
    # see the note above), use_feature_cache=True for test (already warmed).
    train_ds = FiLMDataset(df_train, feature_dirs, gene_cols, clinical_cols,
                           n_tiles=None, deterministic=False, use_feature_cache=False)
    val_ds   = FiLMDataset(df_val,   feature_dirs, gene_cols, clinical_cols,
                           n_tiles=None, deterministic=True, use_feature_cache=False)
    test_ds  = FiLMDataset(df_test,  feature_dirs, gene_cols, clinical_cols,
                           n_tiles=None, deterministic=True, use_feature_cache=True)

    train_loader = DataLoader(train_ds, batch_size=1, shuffle=True,  num_workers=8, pin_memory=True, persistent_workers=True, prefetch_factor=2,)
    val_loader   = DataLoader(val_ds,   batch_size=1, shuffle=False, num_workers=8, pin_memory=True, persistent_workers=True, prefetch_factor=2,)
    # num_workers=0 for test: it's already fully cached, so there's
    # nothing to parallelise, and this avoids any residual risk of a
    # worker privately caching a not-yet-cached test slide.
    test_loader  = DataLoader(test_ds,  batch_size=1, shuffle=False, num_workers=0)

    print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

    # Model, loss, optimiser
    model     = FiLMMILModel(
        feat_dim=1536, n_genes=len(gene_cols), use_film=FILM_ENABLED,
        aggregator=AGGREGATOR, agg_hidden_dim=AGG_HIDDEN_DIM,
        graph_k=GRAPH_K, graph_layers=GRAPH_LAYERS,
        transmil_heads=TRANSMIL_HEADS, transmil_layers=TRANSMIL_LAYERS,
        transmil_dim_feedforward=TRANSMIL_DIM_FEEDFORWARD,
    ).to(device)
    loss_fn   = CompositeLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, patience=2, factor=0.5
    )

    best_val_pcc = -np.inf
    best_weights = None
    patience_ctr = 0

    for epoch in range(1, MAX_EPOCHS + 1):
        train_loss, train_pcc, _, _, _, _ = run_epoch(model, train_loader, loss_fn, optimizer, epoch, device, training=True, panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,)
        val_loss, val_pcc, _, _, _, _ = run_epoch(model, val_loader, loss_fn, optimizer, epoch, device, training=False, panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,)
        scheduler.step(val_loss)

        print(f"  Epoch {epoch:3d} | "
              f"Train loss={train_loss:.4f} PCC={train_pcc:.4f} | "
              f"Val loss={val_loss:.4f} PCC={val_pcc:.4f}")

        if val_pcc > best_val_pcc:
            best_val_pcc = val_pcc
            best_weights = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            # Save best weights to disk immediately in case of disconnect
            torch.save(best_weights, f"{AGG_OUTPUT_DIR}/fold{fold}_best_model.pt")
            patience_ctr = 0
        else:
            patience_ctr += 1
            if patience_ctr >= PATIENCE:
                print(f"  Early stopping at epoch {epoch}")
                break

    # Test set evaluation
    model.load_state_dict(best_weights)
    model.to(device)
    _, _, test_preds, test_labels, test_panel_preds, test_panel_labels = run_epoch(model, test_loader, loss_fn, optimizer, 999, device, training=False, panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,)

    gene_pccs = compute_gene_pccs(test_preds, test_labels)
    apm_pcc   = compute_panel_pcc(test_preds, test_labels, gene_cols, APM_GENES, gene_symbol_to_idx)
    tis_pcc   = compute_panel_pcc(test_preds, test_labels, gene_cols, TIS_GENES, gene_symbol_to_idx)
    apm_auc   = compute_auc(test_preds, test_labels, gene_cols, APM_GENES, gene_symbol_to_idx)
    tis_auc   = compute_auc(test_preds, test_labels, gene_cols, TIS_GENES, gene_symbol_to_idx)

    # Subtype-split evaluation
    luad_mask = np.array([r["subtype"] == "LUAD" for r in test_ds.records])
    lusc_mask = ~luad_mask
    subtype_results = {}
    for name, mask in [("LUAD", luad_mask), ("LUSC", lusc_mask)]:
        if mask.sum() == 0:
            continue
        p, l = test_preds[mask], test_labels[mask]
        subtype_results[name] = {
            "APM_PCC": compute_panel_pcc(p, l, gene_cols, APM_GENES, gene_symbol_to_idx),
            "TIS_PCC": compute_panel_pcc(p, l, gene_cols, TIS_GENES, gene_symbol_to_idx),
            "APM_AUC": compute_auc(p, l, gene_cols, APM_GENES, gene_symbol_to_idx),
            "TIS_AUC": compute_auc(p, l, gene_cols, TIS_GENES, gene_symbol_to_idx),
            "n":       int(mask.sum()),
        }

    fold_result = {
        "fold":         fold,
        "aggregator":   AGGREGATOR,
        "best_val_pcc": float(best_val_pcc),
        "APM_PCC":      float(apm_pcc),
        "TIS_PCC":      float(tis_pcc),
        "APM_AUC":      float(apm_auc),
        "TIS_AUC":      float(tis_auc),
        "gene_pccs":    {g.replace("_fpkm_uq", ""): float(gene_pccs[i])
                         for i, g in enumerate(gene_cols)},
        "subtype":      subtype_results,
    }
    fold_results.append(fold_result)

    print(f"\nFold {fold} Results:")
    print(f"  APM  PCC={apm_pcc:.4f}  AUC={apm_auc:.4f}")
    print(f"  TIS  PCC={tis_pcc:.4f}  AUC={tis_auc:.4f}")
    for name, res in subtype_results.items():
        print(f"  {name} (n={res['n']}): APM={res['APM_PCC']:.4f}, TIS={res['TIS_PCC']:.4f}")

    # Save checkpoint after this fold completes
    save_checkpoint(fold_results, fold)

print("\n All folds complete.")
```

```python
# Cross-validation summary
print("CROSS-VALIDATION SUMMARY")

for metric in ["APM_PCC", "TIS_PCC", "APM_AUC", "TIS_AUC"]:
    vals = [r[metric] for r in fold_results]
    print(f"  {metric:12s}: {np.mean(vals):.4f} ± {np.std(vals):.4f}")

print()
for subtype in ["LUAD", "LUSC"]:
    print(f"  {subtype}")
    for metric in ["APM_PCC", "TIS_PCC", "APM_AUC", "TIS_AUC"]:
        vals = [r["subtype"][subtype][metric]
                for r in fold_results if subtype in r["subtype"]]
        if vals:
            print(f"  {metric:12s}: {np.mean(vals):.4f} ± {np.std(vals):.4f}")

# Gene-level summary — top and bottom 5 across folds
print("\n Gene-level PCC (mean across folds)")
all_gene_symbols = list(fold_results[0]["gene_pccs"].keys())
mean_gene_pccs = {
    g: np.mean([r["gene_pccs"][g] for r in fold_results])
    for g in all_gene_symbols
}
sorted_genes = sorted(mean_gene_pccs.items(), key=lambda x: x[1], reverse=True)
print("  Top 5 genes:")
for g, r in sorted_genes[:5]:
    print(f"    {g:15s}: {r:.4f}")
print("  Bottom 5 genes:")
for g, r in sorted_genes[-5:]:
    print(f"    {g:15s}: {r:.4f}")
```

## Aggregator-Architecture Robustness Sweep (CLAM / TransMIL / Graph-MIL)

**Main technical-development contribution of this notebook section.**

The FiLM contribution above sits on top of one specific tile-pooling
architecture (additive Attention-MIL). To show that the subtype-conditioned
immune-expression finding is a property of the *modelling idea* — and not an
artefact of that one aggregator — we re-run the exact same frozen UNI2-h
tile embeddings, the same 80/20 train-dev/test split, and the same
loss/training schedule through four different MIL pooling architectures,
changing only the aggregator below FiLM/clinical/heads:

| Aggregator | Idea |
|---|---|
| `abmil`    | Additive Attention-MIL (Ilse et al., 2018) — the baseline used above |
| `clam`     | CLAM-style gated-attention pooling (Lu et al., 2021) |
| `transmil` | Compact self-attention / Transformer MIL, TransMIL-style (Shao et al., 2021) |
| `graph`    | k-NN feature-space graph-MIL with residual message passing |

See the docstrings of `CLAMGatedAttentionMIL`, `TransMILAggregator`, and
`GraphMILAggregator` in `train_film_mil.py` for the specific simplifications
made relative to each published architecture (e.g. CLAM's instance-level
clustering loss assumes a discrete class label and has no analogue for our
continuous 35-gene regression targets, so it is omitted; TransMIL here uses
standard — not Nystrom — self-attention, so tile counts are capped via
`AGG_MAX_TILES` below).

This section is independent of, and can be run instead of or after, the
single-aggregator FiLM run above — it reuses the same `df`, `gene_cols`,
`clinical_cols`, `feature_dirs`, `df_dev`, `df_test`, `panel_idx`, and
`panel_gene_fallback_idx` already loaded earlier in this notebook.

```python
# Aggregator sweep configuration
# Edit AGGREGATORS_TO_SWEEP to run a subset (e.g. ["abmil", "transmil"]) if
# you want a faster/cheaper pass first.

AGGREGATORS_TO_SWEEP = ["abmil", "clam", "transmil", "graph"]

# Tiles/slide fed to the aggregator. transmil and graph scale ~O(N^2) in
# tile count (self-attention / kNN similarity matrix), so capping this is
# strongly recommended once bags exceed a few thousand tiles. The SAME cap
# is applied to every aggregator in the sweep (including abmil/clam) so the
# comparison stays apples-to-apples on identical input bags.
AGG_MAX_TILES = 2000

AGG_HIDDEN_DIM = 256          # attention MLP / gated-attention MLP / graph-conv width
GRAPH_K = 8                   # graph aggregator: nearest neighbours per tile node
GRAPH_LAYERS = 2              # graph aggregator: number of residual graph-conv layers
TRANSMIL_HEADS = 8            # transmil aggregator: self-attention heads
TRANSMIL_LAYERS = 2           # transmil aggregator: Transformer encoder layers
TRANSMIL_DIM_FEEDFORWARD = 512

SWEEP_OUTPUT_DIR = f"{OUTPUT_DIR}/aggregator_sweep"
os.makedirs(SWEEP_OUTPUT_DIR, exist_ok=True)

print(f"Aggregators to sweep : {AGGREGATORS_TO_SWEEP}")
print(f"Agg max tiles         : {AGG_MAX_TILES}")
print(f"Sweep output dir      : {SWEEP_OUTPUT_DIR}")
```

```python
# Per-aggregator checkpoint utilities (mirrors the checkpoint utilities used
# above for the single FiLM run, but keyed by aggregator name so each
# aggregator in the sweep can resume independently if the Kaggle session
# disconnects partway through).

def agg_checkpoint_path(aggregator: str) -> str:
    return f"/kaggle/working/checkpoint_agg_{aggregator}.json"

def save_agg_checkpoint(aggregator: str, fold_results: list, completed_fold: int):
    checkpoint = {"completed_fold": completed_fold, "fold_results": fold_results}
    with open(agg_checkpoint_path(aggregator), "w") as f:
        json.dump(checkpoint, f, indent=2, default=_json_default)
    print(f"[{aggregator}] Checkpoint saved after fold {completed_fold}")

def load_agg_checkpoint(aggregator: str):
    path = agg_checkpoint_path(aggregator)
    if Path(path).exists():
        with open(path) as f:
            cp = json.load(f)
        start_fold = cp["completed_fold"] + 1
        print(f"[{aggregator}] Resuming from fold {start_fold} "
              f"({cp['completed_fold']} folds already completed)")
        return cp["fold_results"], start_fold
    return [], 0

print("Per-aggregator checkpoint utilities ready.")
```

```python
# Aggregator-architecture robustness sweep: main training loop
#
# Outer loop over aggregators, inner loop over folds (identical structure to
# the single-aggregator "Main Training" cell above), so that everything
# except the aggregator itself — data split, FiLM setting, loss, epochs,
# patience — is held fixed across the sweep.

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Configure + preload the feature cache once for the whole sweep -- ONLY
# for the fixed test set (see the single-run training cell above for the
# full explanation of why train/val deliberately do NOT use the cache).
configure_feature_cache(max_gb=FEATURE_CACHE_MAX_GB, dtype=FEATURE_CACHE_DTYPE)
_test_paths = [r["h5_path"] for r in
               FiLMDataset(df_test, feature_dirs, gene_cols, clinical_cols).records]
preload_paths(_test_paths)

sweep_fold_results = {}  # {aggregator: [fold_result, ...]}

for aggregator in AGGREGATORS_TO_SWEEP:
    print(f"\n{'#'*70}")
    print(f"# AGGREGATOR SWEEP: {aggregator}")
    print(f"{'#'*70}")

    agg_output_dir = f"{SWEEP_OUTPUT_DIR}/{aggregator}"
    os.makedirs(agg_output_dir, exist_ok=True)

    kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=90)
    fold_results, start_fold = load_agg_checkpoint(aggregator)

    for fold, (train_idx, val_idx) in enumerate(kf.split(df_dev)):

        if fold < start_fold:
            print(f"[{aggregator}] Skipping fold {fold} (already completed)")
            continue

        print(f"\n{'='*60}")
        print(f"[{aggregator}] FOLD {fold} / {N_FOLDS - 1}")
        print(f"{'='*60}")

        df_train = df_dev.iloc[train_idx].reset_index(drop=True)
        df_val   = df_dev.iloc[val_idx].reset_index(drop=True)

        # Datasets — AGG_MAX_TILES caps tiles/slide identically across every
        # aggregator in the sweep (see config cell above).
        # use_feature_cache=False for train/val (bypass the cache -- see the
        # note above), use_feature_cache=True for test (already warmed).
        train_ds = FiLMDataset(df_train, feature_dirs, gene_cols, clinical_cols,
                               n_tiles=AGG_MAX_TILES, deterministic=False,
                               use_feature_cache=False)
        val_ds   = FiLMDataset(df_val,   feature_dirs, gene_cols, clinical_cols,
                               n_tiles=AGG_MAX_TILES, deterministic=True,
                               use_feature_cache=False)
        test_ds  = FiLMDataset(df_test,  feature_dirs, gene_cols, clinical_cols,
                               n_tiles=AGG_MAX_TILES, deterministic=True,
                               use_feature_cache=True)

        train_loader = DataLoader(train_ds, batch_size=1, shuffle=True,  num_workers=4,
                                  pin_memory=True, persistent_workers=True, prefetch_factor=4)
        val_loader   = DataLoader(val_ds,   batch_size=1, shuffle=False, num_workers=2,
                                  pin_memory=True, persistent_workers=True, prefetch_factor=4)
        # num_workers=0 for test: it's already fully cached, so there's
        # nothing to parallelise, and this avoids any residual risk of a
        # worker privately caching a not-yet-cached test slide.
        test_loader  = DataLoader(test_ds,  batch_size=1, shuffle=False, num_workers=0)

        print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

        # Model, loss, optimiser — only the aggregator changes across the sweep.
        model = FiLMMILModel(
            feat_dim=1536, n_genes=len(gene_cols), use_film=FILM_ENABLED,
            aggregator=aggregator, agg_hidden_dim=AGG_HIDDEN_DIM,
            graph_k=GRAPH_K, graph_layers=GRAPH_LAYERS,
            transmil_heads=TRANSMIL_HEADS, transmil_layers=TRANSMIL_LAYERS,
            transmil_dim_feedforward=TRANSMIL_DIM_FEEDFORWARD,
        ).to(device)
        loss_fn = CompositeLoss()
        optimizer = torch.optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-5)
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=2, factor=0.5)

        best_val_pcc = -np.inf
        best_weights = None
        patience_ctr = 0

        for epoch in range(1, MAX_EPOCHS + 1):
            train_loss, train_pcc, _, _, _, _ = run_epoch(
                model, train_loader, loss_fn, optimizer, epoch, device, training=True,
                panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,
            )
            val_loss, val_pcc, _, _, _, _ = run_epoch(
                model, val_loader, loss_fn, optimizer, epoch, device, training=False,
                panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,
            )
            scheduler.step(val_loss)

            print(f"  [{aggregator}] Epoch {epoch:3d} | "
                  f"Train loss={train_loss:.4f} PCC={train_pcc:.4f} | "
                  f"Val loss={val_loss:.4f} PCC={val_pcc:.4f}")

            if val_pcc > best_val_pcc:
                best_val_pcc = val_pcc
                best_weights = {k: v.cpu().clone() for k, v in model.state_dict().items()}
                torch.save(best_weights, f"{agg_output_dir}/fold{fold}_best_model.pt")
                patience_ctr = 0
            else:
                patience_ctr += 1
                if patience_ctr >= PATIENCE:
                    print(f"  [{aggregator}] Early stopping at epoch {epoch}")
                    break

        # Test set evaluation
        model.load_state_dict(best_weights)
        model.to(device)
        _, _, test_preds, test_labels, test_panel_preds, test_panel_labels = run_epoch(
            model, test_loader, loss_fn, optimizer, 999, device, training=False,
            panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,
        )

        gene_pccs = compute_gene_pccs(test_preds, test_labels)
        apm_pcc = compute_panel_pcc(test_preds, test_labels, gene_cols, APM_GENES, gene_symbol_to_idx)
        tis_pcc = compute_panel_pcc(test_preds, test_labels, gene_cols, TIS_GENES, gene_symbol_to_idx)
        apm_auc = compute_auc(test_preds, test_labels, gene_cols, APM_GENES, gene_symbol_to_idx)
        tis_auc = compute_auc(test_preds, test_labels, gene_cols, TIS_GENES, gene_symbol_to_idx)

        if test_panel_preds is not None:
            head_pccs = compute_panel_head_pcc(test_panel_preds, test_panel_labels)
            apm_head_pcc, tis_head_pcc = head_pccs["APM"], head_pccs["TIS"]
        else:
            apm_head_pcc = tis_head_pcc = None

        luad_mask = np.array([r["subtype"] == "LUAD" for r in test_ds.records])
        lusc_mask = ~luad_mask
        subtype_results = {}
        for name, mask in [("LUAD", luad_mask), ("LUSC", lusc_mask)]:
            if mask.sum() == 0:
                continue
            p, l = test_preds[mask], test_labels[mask]
            res = {
                "APM_PCC": float(compute_panel_pcc(p, l, gene_cols, APM_GENES, gene_symbol_to_idx)),
                "TIS_PCC": float(compute_panel_pcc(p, l, gene_cols, TIS_GENES, gene_symbol_to_idx)),
                "APM_AUC": float(compute_auc(p, l, gene_cols, APM_GENES, gene_symbol_to_idx)),
                "TIS_AUC": float(compute_auc(p, l, gene_cols, TIS_GENES, gene_symbol_to_idx)),
                "n": int(mask.sum()),
            }
            if test_panel_preds is not None:
                head_res = compute_panel_head_pcc(test_panel_preds[mask], test_panel_labels[mask])
                res["APM_head_PCC"], res["TIS_head_PCC"] = head_res["APM"], head_res["TIS"]
            subtype_results[name] = res

        fold_result = {
            "fold": fold,
            "aggregator": aggregator,
            "best_val_pcc": float(best_val_pcc),
            "APM_PCC": float(apm_pcc), "TIS_PCC": float(tis_pcc),
            "APM_AUC": float(apm_auc), "TIS_AUC": float(tis_auc),
            "APM_head_PCC": float(apm_head_pcc) if apm_head_pcc is not None else None,
            "TIS_head_PCC": float(tis_head_pcc) if tis_head_pcc is not None else None,
            "gene_pccs": {g.replace("_fpkm_uq", ""): float(gene_pccs[i]) for i, g in enumerate(gene_cols)},
            "subtype": subtype_results,
        }
        fold_results.append(fold_result)

        print(f"\n[{aggregator}] Fold {fold} Results:")
        print(f"  APM  PCC={apm_pcc:.4f}  AUC={apm_auc:.4f}")
        print(f"  TIS  PCC={tis_pcc:.4f}  AUC={tis_auc:.4f}")

        save_agg_checkpoint(aggregator, fold_results, fold)

    sweep_fold_results[aggregator] = fold_results

    # Save this aggregator's full results immediately, mirroring the
    # single-run "Save final results JSON" cell above.
    with open(f"{agg_output_dir}/results.json", "w") as f:
        json.dump(fold_results, f, indent=2, default=_json_default)
    print(f"[{aggregator}] Results saved to {agg_output_dir}/results.json")

print("\n All aggregators complete.")
```

```python
# Aggregator robustness sweep: summary table
#
# Mean +/- std across folds for each aggregator, ready to drop directly
# into a paper table/figure.

import pandas as pd

METRICS = ["APM_PCC", "TIS_PCC", "APM_AUC", "TIS_AUC", "APM_head_PCC", "TIS_head_PCC"]

summary_rows = []
for aggregator, fold_results in sweep_fold_results.items():
    row = {"aggregator": aggregator, "n_folds": len(fold_results)}
    for metric in METRICS:
        vals = [r[metric] for r in fold_results if r.get(metric) is not None]
        if vals:
            row[f"{metric}_mean"] = float(np.mean(vals))
            row[f"{metric}_std"]  = float(np.std(vals))
    summary_rows.append(row)

summary_df = pd.DataFrame(summary_rows).set_index("aggregator")
summary_path = f"{SWEEP_OUTPUT_DIR}/aggregator_sweep_summary.csv"
summary_df.to_csv(summary_path)

with open(f"{SWEEP_OUTPUT_DIR}/aggregator_sweep_summary.json", "w") as f:
    json.dump(summary_rows, f, indent=2, default=_json_default)

print(f"Summary saved to: {summary_path}")
summary_df
```

```python
# Raw predictions + labels per fold (reloaded from checkpoints)
import torch
from torch.utils.data import DataLoader
from pathlib import Path

# Fixed test set — identical across every fold by construction
test_ds = FiLMDataset(df_test, feature_dirs, gene_cols, clinical_cols,
                       n_tiles=None, deterministic=True)
test_loader = DataLoader(test_ds, batch_size=1, shuffle=False,
                          num_workers=2, pin_memory=True)

test_sids_ordered     = [r["sid"] for r in test_ds.records]      # row order guarantee
test_subtypes_ordered = [r["subtype"] for r in test_ds.records]  # (shuffle=False)

all_fold_preds = {}   # fold -> (n_test, n_genes) array
all_fold_labels = None

for fold in range(N_FOLDS):
    # AGG_OUTPUT_DIR / AGGREGATOR (+ hyperparams) come from whichever
    # single-run training cell you ran last -- reconstructing the model
    # with the SAME aggregator is required for load_state_dict to match,
    # since different aggregators are different architectures.
    ckpt_path = Path(AGG_OUTPUT_DIR) / f"fold{fold}_best_model.pt"
    if not ckpt_path.exists():
        print(f"Fold {fold}: checkpoint not found, skipping (not completed yet?)")
        continue

    model = FiLMMILModel(
        feat_dim=1536, n_genes=len(gene_cols), use_film=FILM_ENABLED,
        aggregator=AGGREGATOR, agg_hidden_dim=AGG_HIDDEN_DIM,
        graph_k=GRAPH_K, graph_layers=GRAPH_LAYERS,
        transmil_heads=TRANSMIL_HEADS, transmil_layers=TRANSMIL_LAYERS,
        transmil_dim_feedforward=TRANSMIL_DIM_FEEDFORWARD,
    ).to(device)
    model.load_state_dict(torch.load(ckpt_path, map_location=device))
    model.eval()

    loss_fn   = CompositeLoss()
    optimizer = torch.optim.Adam(model.parameters())  # unused (training=False), needed for signature

    _, _, test_preds, test_labels, test_panel_preds, test_panel_labels = run_epoch(model, test_loader, loss_fn, optimizer, 999, device, training=False, panel_idx=panel_idx, panel_gene_fallback_idx=panel_gene_fallback_idx,)
    all_fold_preds[fold] = test_preds
    all_fold_labels = test_labels  # identical every fold (same fixed test set)

    np.savez(
        f"{EXPORT_DIR}/{AGGREGATOR}_fold{fold}_test_predictions.npz",
        preds=test_preds, labels=test_labels,
        submitter_id=np.array(test_sids_ordered),
        subtype=np.array(test_subtypes_ordered),
        gene_cols=np.array(gene_cols),
    )
    print(f"  Fold {fold}: saved raw predictions {test_preds.shape} "
          f"to {AGGREGATOR}_fold{fold}_test_predictions.npz")

# Ensemble prediction (mean across folds, same fixed test set) — your
# headline number, since all folds evaluate the identical test set
ensemble_preds = np.mean(list(all_fold_preds.values()), axis=0)
np.savez(
    f"{EXPORT_DIR}/{AGGREGATOR}_ensemble_test_predictions.npz",
    preds=ensemble_preds, labels=all_fold_labels,
    submitter_id=np.array(test_sids_ordered),
    subtype=np.array(test_subtypes_ordered),
    gene_cols=np.array(gene_cols),
)
print(f"Ensemble predictions saved ({len(all_fold_preds)} folds averaged)")

# Tidy long-format CSV for quick plotting (predicted vs actual, per gene, per patient)
import pandas as pd
rows = []
for i, sid in enumerate(test_sids_ordered):
    for g_idx, g in enumerate(gene_cols):
        rows.append({
            "submitter_id": sid,
            "subtype": test_subtypes_ordered[i],
            "gene": g.replace("_fpkm_uq", ""),
            "predicted": ensemble_preds[i, g_idx],
            "actual": all_fold_labels[i, g_idx],
        })
pd.DataFrame(rows).to_csv(f"{EXPORT_DIR}/{AGGREGATOR}_ensemble_predictions_long.csv", index=False)
print(f"Long-format CSV for scatter plots saved ({len(rows)} rows)")
```

```python
# Bootstrap CIs + permutation p-values
N_BOOT = 2000
N_PERM = 2000
rng = np.random.default_rng(0)

def bootstrap_ci(preds, labels, gene_list, n_boot=N_BOOT):
    n = preds.shape[0]
    vals = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, size=n)  # resample patients with replacement
        vals.append(compute_panel_pcc(preds[idx], labels[idx], gene_cols, gene_list, gene_symbol_to_idx))
    vals = np.array(vals)
    return float(np.mean(vals)), float(np.percentile(vals, 2.5)), float(np.percentile(vals, 97.5))

def permutation_pvalue(preds, labels, gene_list, n_perm=N_PERM):
    observed = compute_panel_pcc(preds, labels, gene_cols, gene_list, gene_symbol_to_idx)
    n = preds.shape[0]
    null_vals = np.empty(n_perm)
    for i in range(n_perm):
        perm_idx = rng.permutation(n)
        null_vals[i] = compute_panel_pcc(preds, labels[perm_idx], gene_cols, gene_list, gene_symbol_to_idx)
    p = (np.sum(null_vals >= observed) + 1) / (n_perm + 1)
    return float(observed), float(p)

stats_summary = {}
targets = {"APM": APM_GENES, "TIS": TIS_GENES}

# Per-fold + ensemble
sources = {**{f"fold{f}": p for f, p in all_fold_preds.items()}, "ensemble": ensemble_preds}
for name, preds in sources.items():
    stats_summary[name] = {}
    for panel, genes in targets.items():
        mean_pcc, lo, hi = bootstrap_ci(preds, all_fold_labels, genes)
        observed, p = permutation_pvalue(preds, all_fold_labels, genes)
        stats_summary[name][panel] = {
            "PCC": observed, "bootstrap_mean": mean_pcc,
            "CI95_low": lo, "CI95_high": hi, "perm_p": p,
        }
        print(f"{name:10s} {panel}: PCC={observed:.4f} 95%CI=[{lo:.4f},{hi:.4f}] p={p:.4f}")

with open(f"{EXPORT_DIR}/bootstrap_permutation_stats.json", "w") as f:
    json.dump(stats_summary, f, indent=2)
print(f"Saved: {EXPORT_DIR}/bootstrap_permutation_stats.json")
```
